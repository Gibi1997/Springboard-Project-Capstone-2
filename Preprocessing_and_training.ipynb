For the columns 'NAME_EDUCATION_TYPE' and 'NAME_HOUSING_TYPE', few categories are having very less no.s compared to others. So instead of specifying all categories, we can specify these categories together as 'Other'. So that we can limit the no. of columns, when creating dummies.

```python
#To change less occuring categories as 'other'
mask_EDU=df_applicant['NAME_EDUCATION_TYPE'].isin(counts_NAME_EDUCATION_TYPE[counts_NAME_EDUCATION_TYPE<1000].index)
df_applicant['NAME_EDUCATION_TYPE'][mask_EDU]='Other'
print(df_applicant['NAME_EDUCATION_TYPE'].value_counts())
```

```python
#To change less occuring categories as 'other'
mask_HOU=df_applicant['NAME_HOUSING_TYPE'].isin(counts_NAME_HOUSING_TYPE[counts_NAME_HOUSING_TYPE<1000].index)
df_applicant['NAME_HOUSING_TYPE'][mask_HOU]='Other'
print(df_applicant['NAME_HOUSING_TYPE'].value_counts())
```

Machine learning algorithms require their input data to be represented as a vector or a matrix and assume data should be normally distributed. So we have to represent all categories as numbers using Dummy Encoding.

```python
#Creating dummies for category variables
df_applicant_dummies=pd.get_dummies(df_applicant, drop_first =True)
```

```python
#Converting it to int datatype
df_applicant_dummies = df_applicant_dummies.astype(int)
```

```python
df_applicant_dummies.head()
```

```python
#Dropping columns DAYS_BIRTH and DAYS_EMPLOYED to avoid negative no.s by creating two new columns.
df_applicant_dummies['AGE_YEARS'] = (-df_applicant_dummies['DAYS_BIRTH'] / 365).astype(int)
df_applicant_dummies['EMPLOYED_YEARS'] = (-df_applicant_dummies['DAYS_EMPLOYED'] / 365)
df_applicant_dummies.drop(columns=['DAYS_BIRTH','DAYS_EMPLOYED'], inplace=True)
```

```python
#Checking for missing values
print(df_applicant_dummies.isna().sum())
```

There are no missing values in the dataset.

```python
# Drop identifier
df = df_applicant_dummies.drop(columns=['ID'])
```

```python
# Remove duplicate applicants
df = df.drop_duplicates()
```

```python
#Changing the order of STATUS column 
cols = ['STATUS'] + [c for c in df.columns if c != 'STATUS']
df = df[cols]
```

```python
df.head()
```

```python
#Checking class imbalance
df['STATUS'].value_counts(normalize=True)
```

When checked class imbalance, it was found that the dataset is imbalanced. 0.786796 proportion of customers are good customers and others are bad customers.

```python
#Specifying X and y for splitting
X= df.drop("STATUS",axis=1).values
y= df["STATUS"].values
```

```python
#Train test splitting
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state= 42, test_size=0.2, stratify =y)
```

```python
#Standardizing
scaler = StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

All features were standardized for consistency. Binary and dummy variables were retained in standardized form as the subsequent models are scale-invariant.

```python
# save the data to a new csv file
import numpy as np
df.to_csv(
    r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data\Applicant_data_trained.csv',
    index=False
)
np.save(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data\X_train_scaled.npy', X_train_scaled)
np.save(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data\X_test_scaled.npy', X_test_scaled)
np.save(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data\y_train.npy', y_train)
np.save(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data\y_test.npy', y_test)
```
