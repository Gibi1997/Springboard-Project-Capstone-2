```python
#Importing necessary libraries
import os
import math 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#ignore warning messages to ensure clean outputs
import warnings
warnings.filterwarnings('ignore')
```

```python
#Importing the df
#Load the data
df_new= pd.read_csv(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data/Applicant_data_trained.csv')
```

```python
#Importing the splitted data
X_train = np.load(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data/X_train_scaled.npy')
X_test = np.load(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data/X_test_scaled.npy')
y_train = np.load(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data/y_train.npy')
y_test = np.load(r'C:\Users\gibib\Desktop\Data Scientist\Springboard-Project-Capstone-2\Credit Card Data/y_test.npy')
```

This is a classification problem, in supervised learning. Here we have used the following classification models:

Logistic Regression,
K-Nearest Neighbor (KNN),
Decision Trees,
Random Forest &
Gradient Boost

Evaluating the performance of a model by training and testing on the same dataset can lead to the overfitting. Hence the model evaluation is based on splitting the dataset into train and validation set. But the performance of the prediction result depends upon the random choice of the pair of (train,validation) set. Inorder to overcome that, the Cross-Validation procedure is used where under the k-fold CV approach, the training set is split into k smaller sets, where a model is trained using k-1 of the folds as training data and the model is validated on the remaining part.

Classification/ Confusion Matrix: This matrix summarizes the correct and incorrect classifications that a classifier produced for a certain dataset. Rows and columns of the classification matrix correspond to the true and predicted classes respectively. The two diagonal cells (upper left, lower right) give the number of correct classifications, where the predicted class coincides with the actual class of the observation. The off diagonal cells gives the count of the misclassification. The classification matrix gives estimates of the true classification and misclassification rates.

Since this is an imbalanced classification problem, we focuses on f1-score as a metrics.

```python
#Import ML models:

from sklearn.linear_model import LogisticRegression
from sklearn. neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier 
from sklearn.model_selection import cross_val_score, KFold, GridSearchCV
from sklearn.metrics import f1_score, classification_report
```

```python
#For each model, we are performing cross-validation based on f1-scores for training data.
from imblearn.over_sampling import SMOTE
# Handle class imbalance using SMOTE
sm = SMOTE(random_state=42)
X_train_res, y_train_res = sm.fit_resample(X_train, y_train)

# Define models (with class_weight for applicable models)
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, class_weight="balanced", solver="lbfgs"),
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(random_state=42, class_weight="balanced"),
    "Random Forest": RandomForestClassifier(random_state=42, class_weight="balanced"),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)  # SMOTE handles imbalance
}

# Cross-validation setup
kf = KFold(n_splits=6, shuffle=True, random_state=42)
cv_results = {}

# Evaluate each model with F1-macro
for name, model in models.items():
    scores = cross_val_score(
        model,
        X_train_res,
        y_train_res,
        cv=kf,
        scoring="f1_macro"  # Macro F1 gives equal weight to minority class
    )
    cv_results[name] = scores
    print(f"{name}")
    print(f"  Mean F1-macro (CV): {scores.mean():.4f}")
    print(f"  Std  F1-macro (CV): {scores.std():.4f}\n")
```

```python
#Checking test set performance
test_results = {}
for name, model in models.items():
    model.fit(X_train_res, y_train_res)
    y_pred = model.predict(X_test)
    
    test_f1_macro = f1_score(y_test, y_pred, average="macro")
    test_results[name] = test_f1_macro
    
    print(f"{name} Test set F1-macro: {test_f1_macro:.4f}")
    print(classification_report(y_test, y_pred))
```

The modeling stage evaluated five candidate models using cross-validated F1 scores on the training data and F1 scores on the test set. Gradient Boosting and Random Forest achieved the highest performance. 
Best model: Gradient Boosting-
 Highest CV F1-macro
 Best generalization on original test set
 Stable across folds (low CV std)
Second-best model: Random Forest-
 Still strong ensemble
 Slightly worse F1-macro, especially on minority class

Other models (Logistic Regression, KNN, Decision Tree) either underfit or fail to capture minority class well, so they are not recommended for final DSM deployment.

```python
#Visual comparison of models based on both train and test data
models_list = list(test_results.keys())
train_scores = [cv_results[m].mean() for m in models_list]
test_scores = [test_results[m] for m in models_list]

plt.figure(figsize=(8,5))
plt.plot(models_list, train_scores, marker='o', label='Train CV F1')
plt.plot(models_list, test_scores, marker='s', label='Test F1')
plt.title('Model Comparison (Train vs Test F1)')
plt.ylabel('F1 Score')
plt.xticks(rotation=20)
plt.ylim(0,1)
plt.legend()
plt.grid(True)
plt.show()
```

Gradient Boosting can be selected as the best model because it balances training performance and test generalization. Random Forest can be a secondary option, but Gradient Boosting has the best train-test consistency.

PARAMETER TUNING FOR THE BEST 2 MODELS

```python
# Parameter tuning for Gradient Boosting
from sklearn.model_selection import RandomizedSearchCV
gb = GradientBoostingClassifier(random_state=42)

gb_param_grid = {
    'n_estimators': [100, 200],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 5, 7],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 5],
    'subsample': [0.7, 0.8, 1.0],
    'max_features': ['sqrt', 'log2', None]
}

gb_random = RandomizedSearchCV(
    estimator=gb,
    param_distributions=gb_param_grid,
    n_iter=20,
    scoring='f1_macro',
    cv=kf,
    n_jobs=-1,
    verbose=1,
    random_state=42
)

gb_random.fit(X_train_res, y_train_res)
best_gb = gb_random.best_estimator_

print("Best GB params:", gb_random.best_params_)
print("Best GB CV F1-macro:", gb_random.best_score_)
```

```python
# Parameter tuning for Random Forest
# Random Forest with class_weight balanced
rf = RandomForestClassifier(random_state=42, class_weight='balanced')

rf_param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 5, 10],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 5],
    'max_features': ['sqrt', 'log2', None]
}

kf = KFold(n_splits=5, shuffle=True, random_state=42)

rf_random = RandomizedSearchCV(
    estimator=rf,
    param_distributions=rf_param_grid,
    n_iter=20,  # faster than full grid
    scoring='f1_macro',
    cv=kf,
    n_jobs=-1,
    verbose=1,
    random_state=42
)

rf_random.fit(X_train_res, y_train_res)
best_rf = rf_random.best_estimator_

print("Best RF params:", rf_random.best_params_)
print("Best RF CV F1-macro:", rf_random.best_score_)
```

```python
#Threshold tuning to maximize F1-macro
def optimize_threshold(model, X_test, y_test):
    y_prob = model.predict_proba(X_test)[:, 1]  # probability of class 1
    best_thresh = 0.5
    best_f1 = 0

    for t in np.arange(0.1, 0.9, 0.01):
        y_pred = (y_prob >= t).astype(int)
        f1 = f1_score(y_test, y_pred, average='macro')
        if f1 > best_f1:
            best_f1 = f1
            best_thresh = t

    print("Optimal threshold:", best_thresh)
    print("Best F1-macro on test set:", best_f1)
    y_pred_final = (y_prob >= best_thresh).astype(int)
    print(classification_report(y_test, y_pred_final))
    return best_thresh,y_pred_final
```

```python
print("----- Random Forest -----")
best_threshold, y_pred_final = optimize_threshold(best_rf, X_test, y_test)

print("----- Gradient Boosting -----")
best_threshold, y_pred_final = optimize_threshold(best_gb, X_test, y_test)
```

Gradient Boosting has higher F1-macro on the test set, meaning it balances predictions between minority and majority classes slightly better than RF.
So Gradient Boosting can be considered as the final model.
